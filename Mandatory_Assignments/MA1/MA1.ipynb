1. Read the CSV file in Pandas and create a DataFrame named Grc_df. What is the number of 
rows and columns in Grc_df? Print the first 10 and last 10 rows of Grc_df.  

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Here reads the dataset from a CSV file
Grc_df = pd.read_csv("Grocery_dataset.csv")

# Here prints the shape of the dataset
print("Shape:", Grc_df.shape)

# Here displays the first 10 rows of the dataset using "head"
print("First 10 rows:")
display(Grc_df.head(10))

# Here displays the last 10 rows of the dataset using "tail"
print("Last 10 rows:")
display(Grc_df.tail(10))
```

2. Are there any null values in the Grc_df? If yes, then in which columns and how many? 
Finally, handle these null values using any strategy shown during the labs.

```python
#find missing values in the dataset
Grc_df.isna().sum()
```

```python
#For Item_Weight, we follow the Lab 3 numerical-data approach 
# and replace missing values with the median.
Grc_df["Item_Weight"] = Grc_df["Item_Weight"].fillna(
    Grc_df["Item_Weight"].median()
)

Grc_df.isna().sum()
```

```python
# For Outlet_Size we follow the Lab 3 categorical-data approach.
# Lab 3 handles a similar categorical missing-value situation by replacing it with the most common category. 
# First check which is most common.
Grc_df["Outlet_Size"].value_counts()
```

```python
# The Medium is the most common, so we replace the missing values with "Medium"
Grc_df["Outlet_Size"] = Grc_df["Outlet_Size"].fillna("Medium")

# Verify.
Grc_df.isna().sum()
```

3. How many unique Outlet Sizes are there in the Grc_df? Which outlet size is maximum, and 
which is minimum?

```python
# Lab 2 Pandas notebook specifically demonstrates .nunique() and .value_counts() for this type of question.
Grc_df["Outlet_Size"].nunique()

#This gives both the number of unique sizes and how often each appears.
Grc_df["Outlet_Size"].value_counts()
```

4. How many unique Item Fat Content types are in the Grc_df? List them. Do you see any 
issues with the Item Fat Content types? If yes, then handle this issue. 

```python
# Pandas initially sees 5 unique categories, but there are really only 2 categories.
Grc_df["Item_Fat_Content"].nunique()
Grc_df["Item_Fat_Content"].value_counts()
```

```python
# Sorting out the categories to make them consistent, 
# so that "LF" and "low fat" are both "Low Fat", and "reg" is "Regular"
Grc_df["Item_Fat_Content"] = Grc_df["Item_Fat_Content"].replace({
    "LF": "Low Fat",
    "low fat": "Low Fat",
    "reg": "Regular"
})

Grc_df["Item_Fat_Content"].value_counts()
```

Initially there are 5 unique values, but some represent the same category using different labels. After cleaning, there are 2 categories: Low Fat and Regular.

5. Drop the columns having index values of 0, 6 and create a new DataFrame Grc_new_df.

```python
# Remove the columns that are not needed for the analysis. 
# We will remove the first and seventh columns (index 0 and 6).
Grc_new_df = Grc_df.drop(
    Grc_df.columns[[0, 6]],
    axis=1
)

Grc_new_df.head()
```

6.  Using different Supermarket type listed in the column Outlet_Type create two different 
    DataFrames from Grc_new_df. Name these DataFrames as SupType_1 and SupType_2.  

```python
# List the unique values in the Outlet_Type column and their counts
Grc_new_df["Outlet_Type"].value_counts()
```

```python
# Create separate DataFrames for each Outlet_Type
SupType_1 = Grc_new_df[
    Grc_new_df["Outlet_Type"] == "Supermarket Type1"
]

SupType_2 = Grc_new_df[
    Grc_new_df["Outlet_Type"] == "Supermarket Type2"
]
```

7. Using Seaborn (“ggplot style”) create a (2,1) subplot of a box plot showing 5-point 
summary of the column Item_MRP for SupType_1 and SupType_2. Which Outlet Type has 
a higher median MRP? Are there any outliers? 

```python
# Visualize the distribution of Item_MRP for each Outlet_Type using boxplots
plt.style.use("ggplot")

plt.figure(figsize=(10, 8))

plt.subplot(2, 1, 1)
sns.boxplot(x="Item_MRP", data=SupType_1)
plt.title("Supermarket Type1 - Item MRP")

plt.subplot(2, 1, 2)
sns.boxplot(x="Item_MRP", data=SupType_2)
plt.title("Supermarket Type2 - Item MRP")

plt.tight_layout()
plt.show()
```

```python
# Calculate and print the median Item_MRP for each Outlet_Type
print("Supermarket Type1 median:", SupType_1["Item_MRP"].median())
print("Supermarket Type2 median:", SupType_2["Item_MRP"].median())
```

Supermarket Type1 has a higher median Item_MRP than Supermarket Type2. Based on the box plots, there do not appear to be any outliers.

8. Concatenate the DataFrames SupType_1 and SupType_2 to create a new DataFrame 
Grc_Concat_df and sort it in ascending order based on the column Item_Outlet_Sales. 
What is the Outlet_Location_type of the store with the lowest sales? 

```python
# Concatenate the two DataFrames
Grc_Concat_df = pd.concat([SupType_1, SupType_2])

# Sort by Item_Outlet_Sales from lowest to highest

Grc_Concat_df = Grc_Concat_df.sort_values(
    by="Item_Outlet_Sales",
    ascending=True
)

Grc_Concat_df.head()
```

```python
# Find the Outlet_Location_Type associated with the lowest sales
Grc_Concat_df.iloc[0]["Outlet_Location_Type"]
```

The lowest sales outlet is tier 3.

9. Using columns Outlet_Size and Outlet_Location_Type to create a multiple index for 
   Grc_Concat_df. 

```python
#Create the MultiIndex
Grc_Concat_df.set_index(
    ["Outlet_Size", "Outlet_Location_Type"],
    inplace=True
)

Grc_Concat_df.head()
```

10. Cut the column Item_Weight of Grc_Concat_df into 10 buckets and compute each bucket's 
    mean, minimum, maximum, and count. 

```python
# Divide Item_Weight into 10 buckets
Grc_Concat_df["Weight_Bucket"] = pd.cut(
    Grc_Concat_df["Item_Weight"],
    bins=10
)

#Create the groups
weight_groups = Grc_Concat_df.groupby(
    "Weight_Bucket",
    observed=True
)
```

```python
#Mean:
weight_groups["Item_Weight"].mean()
```

```python
#Minimum:
weight_groups["Item_Weight"].min()
```

```python
#Maximum:
weight_groups["Item_Weight"].max()
```

```python
#Count:
weight_groups["Item_Weight"].count()
```
